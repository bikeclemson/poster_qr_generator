```python
import qrcode
from PIL import Image,ImageDraw, ImageFont

import matplotlib.pyplot as plt
import numpy as np
```

```python
def gen_qr(url: str, size: int) -> Image.Image:
    """Create a high-quality QR code resized to the requested pixel size."""
    
    qr = qrcode.QRCode(
        version=None,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=10,
        border=4,
    )
    qr.add_data(url)
    qr.make(fit=True)
    qr_image = qr.make_image(fill_color="black", back_color="white").convert("RGBA")
    return qr_image.resize((size, size), Image.Resampling.LANCZOS)
```

```python
def make_qrpage(id='aa',bg_path='./data/bg.png'):
    print(f'ext_id:{id}')

    DPI = 300
    PAGE_WIDTH = round(8.5 * DPI)   # 2550 pixels
    PAGE_HEIGHT = round(11 * DPI)   # 3300 pixels
    background = Image.open(bg_path).convert("RGBA")
    background = background.resize((PAGE_WIDTH, PAGE_HEIGHT))

    # print(PAGE_WIDTH)
    # print(PAGE_HEIGHT)
    # sz = np.shape(background)
    # id = 'aa'

    qr_image = gen_qr(f'https://bikeclemson.github.io/qr?utm_source=qr&utm_medium=r{id}&utm_campaign=cu',500)

    for row in range(0,5):
        background.alpha_composite(qr_image, (int(2550/2)-600, 50 + row*660))
        background.alpha_composite(qr_image, (int(2550/2)+100, 50 + row*660))

    draw = ImageDraw.Draw(background)

    # print(f'ext_id:{id}')

    # Arial
    font = ImageFont.truetype(
        "/System/Library/Fonts/Supplemental/Arial.ttf",
        70,
    )



    for row in range(0,5):

        draw.text(
            (1000,20 + row*660),              # x, y position
            f"{id}",
            fill="red",
            font=font,
        )

    return background

pg = make_qrpage(id='ab',bg_path='./data/bg.png')

plt.figure(dpi=300)
plt.imshow(pg, cmap="gray")
```

```python
# generate unique ID loop

import string
pages = []
pg_num = 50
for first in string.ascii_lowercase:
    for second in string.ascii_lowercase:
        id = first + second
        print(id,pg_num)
        pg = make_qrpage(id=id,bg_path='./data/bg.png')
        pages.append(pg)
        pg_num = pg_num - 1

        if (pg_num < 0):
            break

# pages[0].save(
#     "./data/out_rack.pdf",
#     save_all=True,
#     append_images=pages[1:],
#     resolution=300,
# )
```
